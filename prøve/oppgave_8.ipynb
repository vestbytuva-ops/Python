Forskjellen mellom hardkoding og mykkoding:

Hardkoding handler om at man ofte gjør ting på en vanskeligere måte, ofte kan en hardkodet kode være lengre, repeterende med verdier uten varaibel navn. En hardkodet kode pleier som regel å være mye lengre enn en mykkode. 

En mykkode handler ofte at man skal gjøre koden så enkelt som mulig "simpler is better". Det går litt ut på at man må tenke i fremtiden med hvordan man skal formulere den så bra som mulig og gjøre den brukbar slik at man kan etterhvert bruke koden senere. Det går også ut på at andre kan bruke koden din uten å bli irritert.

```python
pi = 3.14 #velger å definere pi på starten sånn at jeg slipper å skrive det 3.14 flere ganger, unngår hardkoding. 

sirkel ={
    "s1":4,
    "s2":7,
    "s3":5
}

a1 = pi * sirkel["s1"]**2
a2 = pi * sirkel["s2"]**2
a3 = pi * sirkel["s3"]**2

if a1 > a2 and a1 > a3:
    print(f"{a1} har størst areal")
elif a2 > a3:
    print(f"{a2} har størst areal")
else:
    print(f"{a3} har størst areal")
```
