```python
def areal(radius):
    return 3.14 * radius**2


def omkrets(radius):
    return 2 * 3.14 * radius


def diameter(radius):
    return 2 * radius


# Verdiene i ordboken er radiene til sirklene.
sirkel = {
    "s1": 4,
    "s2": 7,
    "s3": 5,
}

a1 = areal(sirkel["s1"])
a2 = areal(sirkel["s2"])
a3 = areal(sirkel["s3"])

if a1 >= a2 and a1 >= a3:
    print(f"s1 har størst areal: {a1:.2f}")
elif a2 >= a3:
    print(f"s2 har størst areal: {a2:.2f}")
else:
    print(f"s3 har størst areal: {a3:.2f}")

# Bruk de to andre funksjonene på hver sirkel.
for navn, radius in sirkel.items():
    print(f"{navn}: omkrets = {omkrets(radius):.2f}, diameter = {diameter(radius)}")




```
