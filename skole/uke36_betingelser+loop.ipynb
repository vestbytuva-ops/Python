### 1. Logiske operatorer

__Sammenligningsoperatorer.__ Disse operasjonene forteller datamaskinen at den skal ta to verdier (eller variabler) og sammenligne dem. Sammenligningsoperatorene forteller også hvordan disse skal sammenlignes (se tabell under). Resultatet av en sammenligningsoperator er en boolsk verdi.

|Operasjon |Eksempel [kode] |
|:---- |:-----: |
|Er lik |`==` |
|Er ikke lik |`!=` |
|Mindre enn |`<` |
|Større enn |`>` |
|Mindre enn eller lik |`<=` |
|Større enn eller lik |`>=` |

__Logiske operatorer.__ En logisk operator forteller datamaskinen hvordan den skal kombinere flere to ulike sammenligningsoperatorer (se tabell under).

|Operasjon |Eksempel [kode] |
|:---- |:-----: |
|Og (&&) |`and` |
|Eller |`or` |
|Er ikke |`not` |

````python
# Eksempel 1.1 Logiske operatorer

print(10 != 5)
````

### 3. Løkker
En av de store fordelene ved å programmere, er at datamaskinen kan la oss gjenta samme oppgave mange ganger. Mange oppgaver kan være lette å løse, men krever mange gjentakelser. Ved å programmere med løkker kan vi beskrive én gjentakelse, og la datamaskinen ta seg av resten.

Det finnes ulike løkker for forskjellige oppgaver. Vi skal se på de to mest brukte løkkene i Python: for-løkker og while-løkker.

**For-løkker** \
Vi bruker en for-løkke når vi eller programmet vårt vet hvor mange gjentakelser vi ønsker.

<div>
<img src="https://innhold.aunivers.no/image/svg/24408/svg_file/for_font.svg" width="450"/>
</div>

```python
for i in range(10):
    print(i)
```
|```i```|Center|
|:-----:|:----:|
|En variabel. Denne vil forandre verdi for hver gjennomgang av løkka |A |

**While-løkker** \
En while-løkke utfører en bestemt oppgave så lenge en gitt betingelse er sann.

<div>
<img src="https://innhold.aunivers.no/image/svg/24410/svg_file/while_font.svg" width="450"/>

````python
# Eksempel 3.1 For-løkke

print('hei')
print('hei')
print('hei')
print('hei')
print('hei\n\n')

# enklere å skrive det over som en løkke->

for i in range(1,5+1):
    #
    #print(i)
    print('hei')
````

````python
# Eksempel 3.2 While-løkke

x = 0

while x < 10:
    x = x + 1
    print(x)
````

````python
# Oppgave: print 5 'hei' ved å bruke en while løkke
x = 0

while x < 5:
    x = x + 1
    print('hei')
````

````python
# Eksempel 3.3 Break

x = 0
tall = float(input('Skriv ett tall:'))
max = 1000
p = tall

while True:
    x = x + 1
    p = p * 2

    if p > max:
        print(f'Vi må doble tallet {tall} {x} antall ganger før det blir større enn {max}.')
        break
````

````python
# Eksempel 3.4 Løkke gjennom en streng

s = 'blyant'

for i in s:
    print(i)
````

````python
# Eksempel 3.5 Løkke gjennom en liste

l = ['blyant','viskelær','bok','PC']

for i in l:
    print(i)
````

##### **Oppgave 2 *Billettpriser***

Et busselskap ønsker å automatisk beregne pris på billetter ut fra aldersinformasjon og reiserute hentet inn fra kunden. 
Priser for standard billetter:

Rute 1 (By - Sjø): 150,-

Rute 2 (Sjø - Fjell): 250,-

Rute 3 (Fjell - By): 50,-


Aldersrabatter:

0-2: Gratis

2-16: -50 % 

16-65: Standard pris

65 +: -65 %


Skriv et program som gir brukeren riktig pris ut fra disse forutsetningene.



````python
# svar oppgave 2
rute = int(input("Velg rute: (press 1: By - Sjø, 150,-), (press 2: Sjø - Fjell, 250,-), (press 3: Fjell - By, 50,-)"))
 
rute_1 = 150
rute_2 = 250
rute_3 = 50
 
def aldersrabatt(pris, alder):
    if alder <= 2:
        pris = pris * 0
    elif 2 < alder <= 16:
        pris = pris * 0.5
    elif 16 < alder <= 65:
        pris = pris * 1
    else:
        pris = pris * 0.35
    return pris

if rute == 1:
    pris = rute_1
    alder = int(input("Skriv alderen din (uten desimaltall)"))
    pris = aldersrabatt(pris, alder)
    print(f"prisen din for ruten er {pris},-")
    
if rute == 2:
    pris = rute_2
    alder = int(input("Skriv alderen din (uten desimaltall)"))
    pris = aldersrabatt(pris, alder)
    print(f"prisen din for ruten er {pris},-")
    
if rute == 3:
    pris = rute_3
    alder = int(input("Skriv alderen din (uten desimaltall)"))
    pris = aldersrabatt(pris, alder)
    print(f"prisen din for ruten er {pris},-")

#Enklere versjon:

# svar oppgave 2 - forenklet

ruter = {
    1: ("By - Sjø", 150),
    2: ("Sjø - Fjell", 250),
    3: ("Fjell - By", 50)
}

def aldersrabatt(pris, alder):
    if alder <= 2:
        return pris * 0
    elif alder <= 16:
        return pris * 0.5
    elif alder <= 65:
        return pris * 1
    else:
        return pris * 0.35

rute = int(input("Velg rute: (1: By - Sjø, 150,-), (2: Sjø - Fjell, 250,-), (3: Fjell - By, 50,-): "))
navn, grunnpris = ruter[rute]

alder = int(input("Skriv alderen din (uten desimaltall): "))
pris = aldersrabatt(grunnpris, alder)

print(f"Prisen din for ruten {navn} er {pris},-")
````

##### **Oppgave 5 *Karaktergrenser***

Skriv et program som tar inn en total poengsum mellom 0-100 fra bruker. Programmet skal printe til konsoll hvilken karakter poengsummen tilsvarer eller skalaen under. Print-setningen skal også inneholde forklarende tekst, og den poengsummen bruker skrev inn. Hvis bruker skriver en poengsum som er mindre enn 0 eller større enn 100 skal det skrives ut en feilmelding til konsoll.

- F = 0-14 poeng
- E = 15-29 poeng
- D = 30-49 poeng
- C = 50-69 poeng
- B = 70-84 poeng
- A = 85-100 poeng

````python
# svar oppgave 5
poengsum = int(input("Skriv inn poengsummen din."))

if 0 <= poengsum <= 14:
    print(f"Du har fått karakter F for tilsvarende score: {poengsum}")
elif 15 <= poengsum <= 29:
    print(f"Du har fått karakter E for tilsvarende score: {poengsum}")
elif 30 <= poengsum <= 49:
    print(f"Du har fått karakter D for tilsvarende score: {poengsum}")
elif 50 <= poengsum <= 69:
    print(f"Du har fått karakter C for tilsvarende score: {poengsum}")
elif 70 <= poengsum <= 84:
    print(f"Du har fått karakter B for tilsvarende score: {poengsum}")
elif 85 <= poengsum <=100:
    print(f"Du har fått karakter A for tilsvarende score: {poengsum}")
else:
    print(f"Det finnes ikke en karakter for tilsvarende score: {poengsum}")
````
