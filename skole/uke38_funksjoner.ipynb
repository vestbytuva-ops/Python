### 2. Hardkoding og mykkoding

Hardkoding (hardcoding) er innenfor programmering det å innlemme data direkte inn i programmet sitt fremfor å hente dataene en skal arbeide med fra en ekstern fil eller annen kilde. Hardkoding egner seg best for data som ikke skal endres (for eksempel fysiske konstanter). For annen data, som man med fordel kan endre på i fremtiden, foretrekkes mykkoding (sofcoding). Mykkoding er et begrep som refererer til å skaffe en verdi eller funksjon fra en ekstern ressurs: for eksempel fra andre filer, kommandolinjeargumenter, databaser, eller brukerinndata. Mykkoding er det motsatte av hardkoding.

*OBS: Vi kommer tilbake til konseptet om hardkoding og mykkoding når vi skal lære oss å importere ekstern data.*

```python
#Eksempel 2.1 et hardkodet program

#Dette programmet finner omkrets og areal til en sirkel med radius = 4.

radius = 4

areal = 3.14 * 4 **2
omkrets = 2 * 3.14 * 4

print(f'Arealet til sirkelen er {3.14 * 4 **2}')
print(f'Omkretsen til sirkelen er {2 * 3.14 * 4}')
```

```python
#Eksempel 2.2 mykkodet samme program

#Alternativ A (forslag fra timene)
radius = 40 #hentet
pi = 3.14159

raPiUs = pi * radius

areal = raPiUs * radius
omkrets = raPiUs *2

print(f'Arealet til sirkelen er {areal}')
print(f'Omkretsen til sirkelen er {omkrets}')

#Alternativ B (lærer)
### GLOBALE VARIABLER ###
radius = 40 #hentet
pi = 3.14159

### FUNKSJONER ###
def A(r):
    return pi * r**2
def O(r):
    return 2 * pi * r

### UTSKRIFT ###
print(f'Arealet til sirkelen er {A(radius)}')
print(f'Omkretsen til sirkelen er {O(radius)}')
```

### 4. Dokumentasjonsstrenger (docstrings)

I python er det vanlig å skrive det som kalles docstrings for å dokumentere funksjoner og programmer. Docstrings skrives rett inn i funksjonen (f.eks. ved bruk av trippel-anførselstegn ''')

Docstrings vises ved bruk av kommandoen help().

```python
#Eksempel 4.1 hvordan skrive docstrings

def sum(a, b):
    '''
    Funksjonen tar inn to tall, legger dem sammen og returnerer summen.
    
    INPUT
    a = tall (float eller int)
    b = tall (float eller int)

    RETURN
    a + b
    '''
    return a + b
```

```python
#Eksempel 4.2 kommandoen help()

def sum(a, b):
    '''
    Funksjonen tar inn to tall, legger dem sammen og returnerer summen.
    
    INPUT
    a = tall (float eller int)
    b = tall (float eller int)

    RETURN
    a + b
    '''
    return a + b

help(sum)
```

Funksjonen **formatTime** tar inn et antall sekunder som har gått siden midnatt fra bruker, og omgjør det til et klokkeslett på dagen. 

Eksempel:

Programmet under har **tre** feil som gjør at det ikke kjører, og ikke gir riktig resultat i print-setningen. Rett opp programmet.

```python
### 2.4 svar

# Hovedfunksjon for å hente inn data fra bruker, formatere til et klokkeslett, og formatere tallene til print-setningen.
def main():
    
    # Hent inn et antall sekunder fra midnatt fra bruker
    time = int(input('Skriv inn antall sekunder fra midnatt: ')) 
    
    # Bruk hjelpefunksjonen formatTime om antall sekunder til et klokkeslett
    hours,mins,secs = formatTime(time)
    
    # Formater tallene til riktig klokkeslett-format ved hjelp av formater-funksjonen og print resultatet
    print(f'{format(hours)}:{format(mins)}:{format(secs)}')

# Hjelpefunksjon for å formatere tallene til klokkeslett
def formater(time):
    if time < 10:
        return '0'+str(time) # Legger til en null foran et tall hvis det er mindre enn null. Legges inn som tekst for at det skal fungere
    else:
        return str(time) # Hvis tallet ikke er mindre enn null returneres det direkte
    
# Hjelpefunksjon for å omgjøre antall sekunder til timer, minutter og sekunder
def formatTime(seconds):
    hours = seconds//3600
    mins = (seconds%3600)//60
    secs = seconds%60
    
    return hours, mins, secs

main()
```

```python
# Kodesnutt for å legge bokstaver fra ord inn i en liste og sortere dem.
def ord(ord1, ord2):
    ord1 = input("skriv inn ord")
    ord2 = input("skriv inn ord nummer 2")
    if sorted(ord1.lower()) == sorted(ord2.lower()):
        print("anagram")
    else:
        print("ikke en anagram")

ord1 = input("skriv inn ord")
ord2 = input("skriv inn ord nummer 2")

ord(ord1,ord2)

#ord_list = list(ord1)
#ord_list.sort()
#print(ord_list) # Debug
```
