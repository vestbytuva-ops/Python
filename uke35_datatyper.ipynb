### 1. Variabler

Variabler utgjør en viktig del av alle programmer. Vi bruker dem til å ta vare på verdier som vi ønsker å bruke i programmene våre. En variabel har et navn og en verdi. Vi kan bruke den samme variabelen mange ganger, og hvis vi endrer verdien i begynnelsen av et program, så vil verdien endres alle steder der vi bruker variabelen. 

```python
#Eksempel 1.1: Tilordne en variabel

a = 3
a = a + 1

print(a)
```

Du synes kanskje det er rart at vi kan skrive a = a + 1? Hvis det er en matematisk likning, så er jo ikke det riktig. Men denne linja er ikke en likning, den viser det vi kaller en tilordning. Verdien til høyre (a + 1) blir tilordnet variabelen som står til venstre (a). Likhetstegnet sier altså ikke at de to sidene er like i dette tilfellet. Likhetstegnet sier at venstre side settes til å være lik resultatet av høyre side.

##### **Navnkonvensjoner**
Variabelnavn i Python kan inneholde store bokstaver, små bokstaver, tall og understrek (_). Variabelnavn skiller mellom store og små bokstaver. Det betyr at antallbilletter og antallBilletter er forskjellige navn. Vi kan bruke stor bokstav for å skille mellom flere ord i et navn. Denne skrivemåten kalles «camel case», fordi de store bokstavene underveis i navnene kan minne om puklene på en kamel. Vi kan også bruke understrek for å skille mellom ord: antall_billetter.

Et vanlig variabelnavn bør begynne med liten bokstav. Når vi senere skal gi navn til klasser, bruker vi stor forbokstav i navnet, derfor burde ikke vanlige variabler ha navn som begynner med stor bokstav. 

Når du lager variabelnavn, er det lurt å bruke navn som forteller hva variabelen inneholder. Skal du regne ut arealet av et rektangel, er variabelnavnene lengde og bredde bedre enn a og b. Da blir det mye lettere å lese koden, spesielt når variablene brukes flere steder i et program.Det kan være fristende å bruke korte variabelnavn for å spare tid, men det er ingenting i veien med lange variabelnavn. Editoren du bruker, vil hjelpe deg med automatisk fullføring av navnene, og koden blir lettere å lese hvis navnene er informative.

**Restriksjoner.** Variabelnavn kan ikke ha et tall i begynnelsen av navnet, de kan ikke inneholde punktum, og et navn kan ikke inneholde mellomrom, bindestrek og en del andre spesialtegn. Videre er det er ikke anbefalt å bruke bokstavene "æ", "ø" eller "å" i variabelnavn da disse er utilgjengelige på de fleste (ikke-norske) tastaturer. Det er heller ikke anbefalt å bruke 'l' (liten bokstav L), 'O' (stor bokstav o), eller 'I' (stor bokstav i) som navn da disse kan forveksles med nummerene én og null i enkelte skrifttyper.

**Konstanter.** En konstant er en variabel som har en verdi som ikke skal forandres underveis i et program. I noen programmeringsspråk kan vi definere konstanter slik at programmet gir feilmelding hvis vi prøver å gi dem nye verdier. Det er ikke mulig i Python, men vi kan likevel vise at en variabel skal behandles som en konstant ved å bruke store bokstaver i variabelnavnet.



| |Navnkonvensjon (PEP 8) |Eksempel [navn] |Eksempel [kode] |
|:-- |:-----|:----:|:---|
|Funksjoner |små bokstaver /m understrek |foo_bar |`def foo_bar:` |
|Klasser |CamelCase |FooBar |`Class FooBar: `|
|Konstanter |store bokstaver /m understrek |FOO_BAR |`FOO_BAR = baz` |
|Variabler |små bokstaver /m understrek |foo_bar |`foo_bar = baz` |

### 3. Streng, Flyttall og Heltall

*(eng. string, float and integer)*

| |Streng |Flyttal|Heltall|
|:-- |:-----:|:----:|:----:|
|Forklaring |Det som vi tenker på som "vanlig" tekst. |Desimaltall. Må bruke punktum, ikke komma. |Heltall. Tall uten desimaler |
|Eksempel |'Hello world' |-3.14 |5 |

Vi kan gjøre forskjellige ting med de forskjellige datatypene. Under er noen eksempler.

```python
#Eksempel 3.1 Gjøre om mellom ulike datatyper

side = input('Hva er sidelengdene i kvadratet ditt?') #lagrer som streng-datatype

# Gjør om til streng
str()

# Gjør om til flyttall
float()

# Gjør om til heltall
int()

areal = side * side #Her får vi en feilmelding dersom vi ikke omgjør 'side' fra streng til flyttall

print(f'Arealet til kvadratet er {areal}.') #print kommandoen skriver ut til terminalen.
```

```python
#Eksempel 3.2 Metoder // Noen forandringer man kan gjøre på streng

text = 'Velkommen til IT2-faget.'

print(text.upper())

print(text.lower())

print(text.capitalize())

print(text.title())

print(text.replace('I','A'))

print(text.index('m'))

print(text)

# Lengde av tekststreng
print(len(text))

a, b, c = text.split(' ') #deler strengen

print(a + c) #konkatenering
```

```python
#Eksempel 3.3 Matematiske operasjoner på flyt- eller heltall

from math import sqrt
from random import randrange

a = 3.3
b = 6.7

print(a + b)    #addisjon
print(b - a)    #subtraksjon
print(a * b)    #multiplikasjon
print(a / b)    #divisjon
print(b // a)   #heltallsdivisjon
print(b % a)    #modulus
print(sqrt(b))  #kvadratrot

# Avrunding på flyttall
a = 12.3456789

print(round(a,2))                   #avrunder tallet 'a' til 2 desimaler.
print(f'Hei, svaret er {a:.2f}')    #avrunder tallet 'a' til 2 desimaler, og fletter inn i en streng.

print(randrange(1,7))   #tilfeldig heltall fra (og med) 1, til (men ikke med) 7 [dvs. 1 | 2 | 3 | 4 | 5 | 6]

#Mye av den samme matematikken fungerer på komplekse tall
c = 2+1j
d = -1j

print(3+c+d)
```
